# **Install & Imports**

```python
!pip install pypdf

import re

from google.colab import files
from pypdf import PdfReader
```

# **Upload PDF**

```python
uploadedFiles = files.upload()

fileName = list(uploadedFiles.keys())[0]

print("File name :", fileName)
```

# **Read PDF & Inspect Pages**

```python
pdfReader = PdfReader(fileName)

for page in pdfReader.pages:
    pageText = page.extract_text()

    print("Page number          :", page.page_number + 1)
    print("Number of lines      :", len(pageText.split("\n")))
    print("Number of words      :", len(pageText.split()))
    print("Number of characters :", len(pageText))
    print("Content              :\n\n", pageText)
```

# **Extract & Clean Text**

```python
text = ""

for page in pdfReader.pages:
    text += page.extract_text()

cleanedText = re.sub(r"\s+", " ", text)

print("Number of characters :", len(cleanedText))
print("Cleaned Content      :\n", cleanedText)
```

# **Chunk Text**

```python
chunkSize = 50

chunks = []

while cleanedText:
    chunks.append(cleanedText[:chunkSize])
    cleanedText = cleanedText[chunkSize:]

print("Chunks:", chunks)
print("Number of chunks:", len(chunks))
```
